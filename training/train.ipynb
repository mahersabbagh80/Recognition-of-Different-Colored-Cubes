# YOLOv5 Cube Detection — Optional Fine-Tune (Local GPU)

Run on dev PC with **NVIDIA RTX 4070 Ti** (CUDA + PyTorch).

**Default path:** Download pretrained `best.pt` from Roboflow Universe — no training required.

Use this notebook **only if** standalone inference on robot camera images fails the accuracy target (see `docs/milestones.md` step 6).

## Fine-tune steps

1. Download Roboflow dataset (YOLOv5 format) from the Universe project
2. Clone YOLOv5 (Ultralytics)
3. Fine-tune YOLOv5s — **20–30 epochs**, classes: `red_cube`, `green_cube`, `blue_cube`
4. Validate on hold-out set
5. Export `best.pt` → `best.onnx`, then repeat TensorRT conversion on Jetson

**Dataset:** https://universe.roboflow.com/jakub-slof/red-green-blue-cube-detection/dataset/1
